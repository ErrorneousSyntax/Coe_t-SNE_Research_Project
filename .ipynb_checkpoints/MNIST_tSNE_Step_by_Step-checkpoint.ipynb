# t-SNE, one iteration at a time

**300 real MNIST training images · 30 per digit · seed 42**

Run the cells from top to bottom. The data subset is embedded in this notebook, so running it needs no dataset download. Each image becomes one dot. Its pixels stay fixed while its two map coordinates move.

We will inspect images → compute fixed original affinities **P** → initialise coordinates **Y** → calculate map affinities **Q** → calculate one scalar **KL loss** → differentiate with respect to every coordinate → update all dots together → repeat.

This teaching implementation uses exact pairwise calculations and ordinary gradient descent with backtracking. We deliberately omit early exaggeration, momentum, and approximate neighbours so that every accepted update reduces the same ordinary KL objective. Production t-SNE commonly uses those additional techniques. This run is an explanation of the mechanics, and its layout may differ from library defaults.

**Interactive output:** the last cell contains an iteration slider and image hover viewer. Trust the notebook in Jupyter and run that cell to enable its JavaScript. Static plots also show saved checkpoints.

```python
# Run this cell only if packages are missing, then restart the kernel.
# %pip install numpy pandas matplotlib ipython
```

```python
import base64
import io
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML, display

SEED = 42
PERPLEXITY = 20.0
MAX_ITERATIONS = 400
INITIAL_STEP_SIZE = 100.0
GRADIENT_TOLERANCE = 1e-7
SAVE_EVERY = 10
plt.rcParams.update({'figure.dpi': 110, 'axes.spines.top': False, 'axes.spines.right': False})
```

## 1. Inspect the real images

The archive below contains a fixed balanced subset from the official MNIST training split. `original_ids` are zero-based row numbers in that split. Labels are used only for displaying images and colours.

Flatten 28 × 28 pixels into 784 features and divide by 255. We keep all 784 features and use Euclidean distances directly, so the relationship to the pixels remains visible.

```python
# Embedded MNIST subset, supplied with the notebook.
encoded_mnist = ''
with np.load(io.BytesIO(base64.b64decode(encoded_mnist)), allow_pickle=False) as archive:
    images = archive['images']
    labels = archive['labels']
    original_ids = archive['ids']
X = images.reshape(len(images), -1).astype(np.float64) / 255.0
N = len(X)
assert X.shape == (300, 784)
print(f"X shape: {X.shape}; {N} images, each with 784 fixed features")
print('Images per digit:', np.bincount(labels))

fig, axes = plt.subplots(2, 10, figsize=(13, 3.4))
for position, ax in enumerate(axes.flat):
    ax.imshow(images[position], cmap='gray', vmin=0, vmax=255)
    ax.set_title(f"digit {labels[position]} | row {position}", fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## 2. Compare original pixel vectors

`squared_distances[i, j]` is the sum of the squared pixel differences between two images. A low distance means their pixel patterns are similar. We exclude an image itself when inspecting its neighbours.

```python
def pairwise_squared_distances(values):
    squared_norms = np.sum(values * values, axis=1)
    distances = squared_norms[:, None] + squared_norms[None, :] - 2 * values @ values.T
    distances = np.maximum(distances, 0.0)
    np.fill_diagonal(distances, 0.0)
    return distances

squared_distances = pairwise_squared_distances(X)
FOCUS = 0  # Change to any subset row from 0 to N-1.
ordered_neighbours = np.argsort(squared_distances[FOCUS])
ordered_neighbours = ordered_neighbours[ordered_neighbours != FOCUS][:5]
fig, axes = plt.subplots(1, 6, figsize=(10, 2.3))
for ax, index in zip(axes, np.r_[FOCUS, ordered_neighbours]):
    ax.imshow(images[index], cmap='gray', vmin=0, vmax=255)
    ax.set_title(f"row {index}, digit {labels[index]}\nd² = {squared_distances[FOCUS, index]:.2f}", fontsize=9)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## 3. Build P once, before any dots move

For image i, Gaussian weights become conditional probabilities:

$$p_{j|i}=\frac{\exp(-d_{ij}^2/(2\sigma_i^2))}{\sum_{k\ne i}\exp(-d_{ik}^2/(2\sigma_i^2))}.$$

We search for each image's Gaussian width so its conditional distribution has the chosen perplexity. With natural logs, entropy is $H_i=-\sum_j p_{j|i}\log p_{j|i}$ and perplexity is $\exp(H_i)$.

Then symmetrise using $p_{ij}=(p_{j|i}+p_{i|j})/(2N)$. Every conditional row sums to one; the **whole joint P matrix** sums to one. P remains fixed throughout all embedding iterations.

```python
def original_affinities(distances, perplexity):
    count = len(distances)
    if not 1 < perplexity < count - 1:
        raise ValueError('Use a perplexity strictly between 1 and N-1.')
    conditional = np.zeros_like(distances)
    sigmas = np.zeros(count)
    achieved = np.zeros(count)
    target_entropy = np.log(perplexity)

    for i in range(count):
        neighbours = np.arange(count) != i
        row = distances[i, neighbours]
        beta = 1.0  # beta = 1 / (2 sigma²)
        lower_beta, upper_beta = 0.0, np.inf
        for search_iteration in range(100):
            # Subtracting the smallest distance prevents all weights underflowing.
            weights = np.exp(-beta * (row - row.min()))
            probabilities = weights / weights.sum()
            positive = probabilities > 0
            entropy = -np.sum(probabilities[positive] * np.log(probabilities[positive]))
            difference = entropy - target_entropy
            if abs(difference) < 1e-8:
                break
            if difference > 0:
                lower_beta = beta
                beta = 2 * beta if np.isinf(upper_beta) else (beta + upper_beta) / 2
            else:
                upper_beta = beta
                beta = (lower_beta + beta) / 2
        if abs(difference) >= 1e-6:
            raise RuntimeError(f'Perplexity search did not converge for row {i}.')
        conditional[i, neighbours] = probabilities
        sigmas[i] = np.sqrt(1 / (2 * beta))
        achieved[i] = np.exp(entropy)

    joint = (conditional + conditional.T) / (2 * count)
    return joint, conditional, sigmas, achieved

P, conditional_P, sigmas, achieved_perplexities = original_affinities(squared_distances, PERPLEXITY)
assert np.allclose(conditional_P.sum(axis=1), 1)
assert np.allclose(P, P.T)
assert np.isclose(P.sum(), 1)
assert np.all(np.diag(P) == 0)
print('Joint P sum:', P.sum())
print('Achieved perplexity range:', achieved_perplexities.min(), achieved_perplexities.max())
display(pd.DataFrame({
    'neighbour_row': ordered_neighbours,
    'digit': labels[ordered_neighbours],
    'squared_pixel_distance': squared_distances[FOCUS, ordered_neighbours],
    'conditional p(j|i)': conditional_P[FOCUS, ordered_neighbours],
    'joint p(i,j)': P[FOCUS, ordered_neighbours],
}))
```

## 4. Put every image at an initial 2D position

Y has shape N × 2. Column 0 is horizontal and column 1 is vertical. We choose seeded random starting positions with standard deviation 0.01 for this ordinary gradient-descent demonstration. Image labels have no role in choosing or updating these coordinates.

```python
rng = np.random.default_rng(SEED)
Y_initial = rng.normal(0, 0.01, size=(N, 2))
Y_initial -= Y_initial.mean(axis=0)

def show_embedding(coordinates, title, ax=None):
    if ax is None:
        _, ax = plt.subplots(figsize=(6, 5))
    dots = ax.scatter(coordinates[:, 0], coordinates[:, 1], c=labels,
                      cmap='tab10', vmin=-0.5, vmax=9.5, s=16, alpha=0.8)
    ax.set(title=title, xlabel='Horizontal coordinate', ylabel='Vertical coordinate')
    ax.set_aspect('equal', adjustable='datalim')
    return dots

dots = show_embedding(Y_initial, 'Iteration 0: random positions')
plt.colorbar(dots, ticks=range(10), label='Digit label, used for display')
plt.show()
```

## 5. Turn positions into Q, and calculate one scalar KL loss

$$w_{ij}=\frac{1}{1+\|y_i-y_j\|^2},\quad Z=\sum_{k\ne l}w_{kl},\quad q_{ij}=w_{ij}/Z.$$

$$L=\sum_{i\ne j}p_{ij}\log(p_{ij}/q_{ij}).$$

Both directions of each pair are included. Self-pairs are excluded. An individual term can be negative; the total KL is nonnegative. Moving one dot changes Z and therefore changes every normalised q, even for pairs whose endpoints stayed still.

```python
def map_affinities(coordinates):
    distances = pairwise_squared_distances(coordinates)
    weights = 1.0 / (1.0 + distances)
    np.fill_diagonal(weights, 0.0)
    normalizer = weights.sum()
    return weights / normalizer, weights, normalizer

def kl_loss(original_probabilities, map_probabilities):
    positive = original_probabilities > 0
    return np.sum(original_probabilities[positive] * (
        np.log(original_probabilities[positive]) - np.log(map_probabilities[positive])))

Q_initial, W_initial, Z_initial = map_affinities(Y_initial)
initial_loss = kl_loss(P, Q_initial)
print(f'Z = {Z_initial:.4f}, sum(Q) = {Q_initial.sum():.6f}, KL = {initial_loss:.6f}')
display(pd.DataFrame({
    'neighbour_row': ordered_neighbours,
    'p': P[FOCUS, ordered_neighbours],
    'q': Q_initial[FOCUS, ordered_neighbours],
    'p log(p/q)': P[FOCUS, ordered_neighbours] * np.log(
        P[FOCUS, ordered_neighbours] / Q_initial[FOCUS, ordered_neighbours]),
}))
```

## 6. Differentiate with respect to both coordinates of every dot

$$\nabla_{y_i} L=4\sum_{j\ne i}(p_{ij}-q_{ij})w_{ij}(y_i-y_j).$$

`gradient[i, 0]` is the derivative of the **total** loss with respect to dot i's horizontal coordinate. `gradient[i, 1]` is its vertical derivative. We compute all rows from the same current positions.

The implementation below is the same sum in matrix form. Multiplying each difference by its coefficient and summing gives `sum(coefficients) * y_i - sum(coefficients * y_j)`.

```python
def kl_gradient(original_probabilities, coordinates, map_probabilities, weights):
    coefficients = (original_probabilities - map_probabilities) * weights
    return 4.0 * (coefficients.sum(axis=1, keepdims=True) * coordinates
                  - coefficients @ coordinates)

initial_gradient = kl_gradient(P, Y_initial, Q_initial, W_initial)
# Independently expand the pair-by-pair sum for our focus image.
focus_pair_gradients = 4 * ((P[FOCUS] - Q_initial[FOCUS]) * W_initial[FOCUS])[:, None] * (
    Y_initial[FOCUS] - Y_initial)
assert np.allclose(focus_pair_gradients.sum(axis=0), initial_gradient[FOCUS])
display(pd.DataFrame({
    'row': np.arange(5),
    'u': Y_initial[:5, 0], 'v': Y_initial[:5, 1],
    'dL/du': initial_gradient[:5, 0], 'dL/dv': initial_gradient[:5, 1],
}))
```

## 7. Verify the derivative numerically

Move just one coordinate a tiny amount in each direction. The slope estimated from the two losses should agree with the analytical gradient. This tests the normalisation and factor of four as well as the chain rule. We check a spread-out configuration to avoid tiny initial derivatives masking errors.

```python
check_positions = np.random.default_rng(7).normal(size=(N, 2))
check_Q, check_W, _ = map_affinities(check_positions)
analytical = kl_gradient(P, check_positions, check_Q, check_W)
epsilon = 1e-5
checks = []
for row, axis in [(0, 0), (0, 1), (17, 0), (71, 1)]:
    plus, minus = check_positions.copy(), check_positions.copy()
    plus[row, axis] += epsilon
    minus[row, axis] -= epsilon
    numerical = (kl_loss(P, map_affinities(plus)[0]) -
                 kl_loss(P, map_affinities(minus)[0])) / (2 * epsilon)
    checks.append([row, axis, analytical[row, axis], numerical])
    assert np.isclose(analytical[row, axis], numerical, atol=1e-8, rtol=1e-4)
display(pd.DataFrame(checks, columns=['row', 'axis (0=u, 1=v)', 'analytical', 'finite difference']))
print('All gradient checks passed.')
```

## 8. Take ONE simultaneous update

$$Y_{\mathrm{new}}=Y-\eta\nabla L(Y).$$

All coordinates move using gradients calculated at the old positions. The line search starts at the chosen step size and halves it if necessary until the ordinary KL loss decreases sufficiently. Trial steps are not accepted iterations. Recentring removes an irrelevant translation and preserves every pair distance.

```python
def take_step(original_probabilities, coordinates, gradient, old_loss, step_size):
    squared_gradient_norm = np.sum(gradient * gradient)
    for backtrack in range(30):
        candidate = coordinates - step_size * gradient
        candidate -= candidate.mean(axis=0)
        candidate_Q, _, _ = map_affinities(candidate)
        candidate_loss = kl_loss(original_probabilities, candidate_Q)
        if candidate_loss <= old_loss - 1e-4 * step_size * squared_gradient_norm:
            return candidate, candidate_loss, step_size, backtrack
        step_size *= 0.5
    raise RuntimeError('Line search failed. Inspect the gradient and step size.')

Y_one, loss_one, accepted_step, backtracks = take_step(
    P, Y_initial, initial_gradient, initial_loss, INITIAL_STEP_SIZE)
print(f'One update: KL {initial_loss:.6f} → {loss_one:.6f}; step size {accepted_step}; backtracks {backtracks}')
display(pd.DataFrame({
    'old u': Y_initial[:5, 0], 'old v': Y_initial[:5, 1],
    'dL/du': initial_gradient[:5, 0], 'dL/dv': initial_gradient[:5, 1],
    'new u': Y_one[:5, 0], 'new v': Y_one[:5, 1],
}))
fig, ax = plt.subplots(figsize=(6, 5))
show_embedding(Y_initial, 'Arrows show the first simultaneous update', ax)
ax.quiver(Y_initial[:, 0], Y_initial[:, 1], Y_one[:, 0] - Y_initial[:, 0],
          Y_one[:, 1] - Y_initial[:, 1], angles='xy', scale_units='xy', scale=1,
          color='black', alpha=0.3, width=0.002)
plt.show()
```

## 9. HERE IS THE ITERATION LOOP

This cell restarts from iteration 0 every time you run it. Within each cycle, calculate Q → calculate the gradient → update every dot → record the new loss. P and the pixels remain fixed. The progress messages report accepted steps. We stop at the gradient tolerance or the iteration limit.

```python
Y = Y_initial.copy()
history = [{'iteration': 0, 'KL': initial_loss, 'gradient_norm': np.linalg.norm(initial_gradient),
            'step_size': np.nan, 'backtracks': 0}]
snapshots = {0: Y.copy()}
stop_reason = 'Maximum iteration count reached'

for iteration in range(1, MAX_ITERATIONS + 1):
    # A. Compare the fixed original probabilities with the CURRENT map.
    Q, weights, _ = map_affinities(Y)
    current_loss = kl_loss(P, Q)

    # B. Calculate both coordinate derivatives for ALL dots before moving any.
    gradient = kl_gradient(P, Y, Q, weights)
    if np.linalg.norm(gradient) < GRADIENT_TOLERANCE:
        stop_reason = 'Gradient norm reached tolerance'
        break

    # C. Move ALL dots together. The image data and P do not change.
    Y, new_loss, step_size, backtracks = take_step(
        P, Y, gradient, current_loss, INITIAL_STEP_SIZE)

    # D. Record measurements at the NEW positions.
    new_Q, new_weights, _ = map_affinities(Y)
    new_gradient = kl_gradient(P, Y, new_Q, new_weights)
    history.append({'iteration': iteration, 'KL': new_loss,
                    'gradient_norm': np.linalg.norm(new_gradient),
                    'step_size': step_size, 'backtracks': backtracks})
    if iteration % SAVE_EVERY == 0 or iteration == 1:
        snapshots[iteration] = Y.copy()
    if iteration % 50 == 0 or iteration == 1:
        print(f'Iteration {iteration:3d} | KL {new_loss:.6f} | gradient norm {np.linalg.norm(new_gradient):.6f}')

history = pd.DataFrame(history)
last_iteration = int(history.iloc[-1]['iteration'])
snapshots[last_iteration] = Y.copy()
assert np.all(np.diff(history['KL']) <= 1e-10)
assert np.isfinite(Y).all()
print(stop_reason)
print(f'Accepted updates: {last_iteration}; KL {initial_loss:.6f} → {history.iloc[-1]["KL"]:.6f}')
```

## 10. Inspect what changed

The loss belongs to the whole embedding. Each coordinate has its own derivative. The checkpoints below have independent axis limits to make the early configurations readable; use the coordinate ticks to compare scales. Colours show digit labels, which the optimiser never receives. A visible island alone does not establish a genuine population or a new digit class.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
axes[0].plot(history['iteration'], history['KL'])
axes[0].set(xlabel='Accepted iteration', ylabel='Total KL loss')
axes[1].semilogy(history['iteration'], history['gradient_norm'])
axes[1].set(xlabel='Accepted iteration', ylabel='Full gradient norm')
plt.tight_layout()
plt.show()

saved_iterations = sorted(snapshots)
chosen = [saved_iterations[i] for i in np.linspace(0, len(saved_iterations)-1, 4, dtype=int)]
fig, axes = plt.subplots(1, 4, figsize=(15, 3.6))
for checkpoint, ax in zip(chosen, axes):
    show_embedding(snapshots[checkpoint], f'Iteration {checkpoint}', ax)
plt.tight_layout()
plt.show()

final_Q, _, _ = map_affinities(Y)
display(pd.DataFrame({
    'neighbour_row': ordered_neighbours,
    'digit': labels[ordered_neighbours],
    'fixed p': P[FOCUS, ordered_neighbours],
    'initial q': Q_initial[FOCUS, ordered_neighbours],
    'final q': final_Q[FOCUS, ordered_neighbours],
}))
```

## 11. Scrub through iterations and hover over images

Move the slider to a saved iteration. Hover near a dot to see its actual MNIST image, label, subset row, and original training ID. Turn off autoscaling to keep the same axes across frames. The saved frames are every ten iterations, plus the first update and final frame.

If the viewer is blank, trust the notebook and rerun this cell in Jupyter. The static outputs above remain available in viewers that block JavaScript.

## What to try next

1. Change `FOCUS` to inspect a different image and its strongest original neighbours.
2. Change `PERPLEXITY` to 5 or 40 and rerun from the P calculation onward, using the same initial coordinates. P changes, so KL values across these different objectives are not directly comparable as quality scores.
3. Change only the random seed and rerun the coordinate initialisation onward. P stays fixed, so the final KL objectives are comparable.

The direct implementation costs O(N²) time and memory for pairwise operations. Keep the embedded subset small while learning. Increasing the iteration limit may refine the layout, but small gradients do not guarantee a global optimum or faithful preservation of every neighbourhood.
